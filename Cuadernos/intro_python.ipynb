### EJEMPLO: Calculo de Metricas de Marketing Digital
Realizado por Daniela(Tutora)

````python
campana = "Black_Friday_2026"       # str: Identificador
impresiones = 150000                # int: Veces que se mostró el anuncio
clics = 4250                        # int: Clics obtenidos
inversion_raw = "1250.50"           # str: Dato crudo exportado de una plataforma

# 1. Transformación de tipo (Casting) para poder operar
inversion_eur = float(inversion_raw)

# 2. Cálculo de KPIs de negocio
ctr = (clics / impresiones) * 100                  # Click-Through Rate (%)
cpc = round(inversion_eur / clics, 2)              # Coste por Clic (€)
es_rentable = ctr > 2.5 and cpc < 0.35             # bool: Regla de negocio

print(f"Campaña: {campana} | CTR: {ctr:.2f}% | CPC: {cpc}€ | ¿Objetivo cumplido?: {es_rentable}")
````

## Ejercicio Calculadora de indices de marketing
Una empresa ha realizado una campaña digital y quiere conocer sus resultados.

**Datos**:
- Impresiones: 12.000
- Clics: 420
- Inversión: 210 €
- Ingresos generados: 680 €

**Calcula**:
- CTR: porcentaje de impresiones que se convierten en clics.
- CPC: coste medio de cada clic.
- Beneficio: ingresos menos inversión.
- ROI: retorno de la inversión en porcentaje.

````python
imp = 12000
clics = 420
inver = 210
inge_gen = 680

ctr = round(((clics/imp)*100), 2)
cpc = (inver/clics)
beneficio = inge_gen - inver
roi = round(((beneficio/inver)*100), 2)

print("== Calculos ===")
print(f"CTR: {ctr}%")
print("CPC:", cpc)
print("Beneficio conseguido:", beneficio)
print(f"ROI: {roi}%")
````

## Ejercicio 2 de Listas

 Una tienda ha registrado estas ventas semanales:

```python
ventas = [1200, 950, 1430, 1100, 1600]
```

Calcula:

- El total de ventas.
- La media(promedio).
- La venta más alta.
- La venta más baja.
- Añade una nueva venta de 1.750 € y vuelve a calcular el total.

````python
ventas = [1200, 950, 1430, 1100, 1600]

total = sum(ventas)
media = sum(ventas)/len(ventas)
mini = min(ventas)
maxi = max(ventas)
print(f"Ventas registradas: {ventas}.")
ventas.append(1750)
total_n = sum(ventas)

print(
    f"Total inicial: {total}.\n"
    f"Media: {media:.2f}.\n"
    f"Venta mínima: {mini}.\n"
    f"Venta máxima: {maxi}.\n"
    f"Indicativo: las ventas han aumentado.\n"
    f"Ventas actualizadas: {ventas}\n"
    f"Total actualizado: {total_n}"
)
````

## Ejercicio de Condicionales

Una empresa recibe valoraciones de sus clientes entre 1 y 5.
Crea una variable llamada satisfaccion y utiliza condicionales para clasificarla:
- 4,5 o más: Excelente.
- 3,5 o más: Buena.
- 2,5 o más: Mejorable.
- Menos de 2,5: Baja.

Prueba el programa con diferentes valores.

````python
sat = round(float(input("Introduzca la satisfacción (0 - 5)")), 1)

print(f"Satisfacción recibida: {sat}/5 estrellas")

if(sat >= 4.5):
    print("Excelente")
elif(sat >= 3.5):
    print("Buena")
elif(sat >= 2.5):
    print("Mejorable")
else:
    print("Baja")
````
