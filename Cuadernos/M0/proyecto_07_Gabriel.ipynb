# Analizador Básico de Sentimiento y Frecuencia en Reseñas (NLP)

### OBJETIVO DE NEGOCIO
Procesar opiniones escritas por clientes para medir el nivel general de satisfacción y detectar las
palabras que más se repiten

### QUÉ DEBE HACER TU CÓDIGO (GUÍA DE TRABAJO)
 - **Normalizar texto**: Convertir cada reseña a minúsculas, eliminar signos de puntuación (., ,, !, :) y dividir las
frases en palabras sueltas.

 - **Clasificar sentimiento**: Comparar las palabras de cada reseña contra las listas de referencia
(palabras_positivas y palabras_negativas) para etiquetar la opinión como Positiva, Negativa o
Neutra/Mixta.

- **Contar frecuencias**: Construir un diccionario que cuente cuántas veces aparece cada palabra clave en todo
el conjunto de reseñas.

```python
# Dataset simulado
palabras_positivas = ["excelente", "rapido", "bueno", "genial", "recomiendo", "perfecto", "amable"]

palabras_negativas = ["lento", "malo", "pesimo", "roto", "caro", "demora", "horrible"]

resenas = [
"El envio fue super rapido y el producto llego perfecto!",
"Pesimo servicio, el paquete llego roto y con mucha demora.",
"Buen producto, aunque un poco caro para lo que ofrece.",
"Excelente atencion, el personal fue muy amable. Lo recomiendo!",
"Horrible experiencia: lento, malo y nadie responde.",
"El producto es bueno y rapido de instalar, pero el soporte es lento.",
"Todo genial, volveria a comprar sin dudarlo."
]
```

**Normalizar texto**: Convertir cada reseña a minúsculas, eliminar signos de puntuación (., ,, !, :) y dividir las
frases en palabras sueltas.

```python
res_pro = []
palabras = []

'''
    Una forma de hacerlo en una sola línea seria mediante:
    i = i = i.lower().translate(str.maketrans("", "", ".,!:"))
'''
for i in resenas:
    # Establezco la reseña en minusculas
    i = i.lower()
    
    # Extraigo los signos de puntuación
    i = i.strip(".")
    i = i.strip(",")
    i = i.strip("!")
    i = i.strip(":")
    
    # Divido las reseñas en palabras y las almaceno como una lista de palabras
    palabras = i.split()
    
    # Almaceno la lista de palabras sueltas en una sola lista
    res_pro.append(palabras)

# Muestro la lista de listas de palabras divididas
for i in res_pro:
    print(i)
```

 - **Clasificar sentimiento**: Comparar las palabras de cada reseña contra las listas de referencia
(palabras_positivas y palabras_negativas) para etiquetar la opinión como Positiva, Negativa o
Neutra/Mixta.

```python
cont = 0
clasi = []

# Recorro cada entrada de la lista de reseñas
for i in res_pro:
    # Recorro cada palabra de la reseña
    for j in i:
        # Recorro cada palabra en palabras_positivas
        for k in palabras_positivas:
            # Comparo con la palabra actual
            if(j == k):
                cont += 1
        # Recorro cada palabra en palabras_negativas
        for k in palabras_negativas:
            # Comparo con la palabra actual
            if(j == k):
                cont -= 1
    # Si hay mayoria de palabras positivas
    if(cont > 0):
        clasi.append("Positiva")
    # Si no hay palabras positivas o negativas, o hay empate
    if(cont == 0):
        clasi.append("Neutra/Mixta")
    # Si hay mayoria de palabras negativas
    else:
        clasi.append("Negativa")

print(clasi)
```

- **Contar frecuencias**: Construir un diccionario que cuente cuántas veces aparece cada palabra clave en todo
el conjunto de reseñas.

```python
dic = {}

# Recorro cada entrada de la lista de reseñas
for i in res_pro:
    # Recorro cada palabra de la reseña
    for j in i:
        # Recorro cada palabra en palabras_positivas
        for k in palabras_positivas:
            # Comparo con la palabra actual
            if(j == k):
                # Reviso si ya existe registro de la palabra, si lo hay suma
                if(k in dic):
                    dic[k] += 1
                # Si la palabra no está, se añade con el contador a 1
                else:
                    dic[k] = 1
        # Recorro cada palabra en palabras_negativas
        for k in palabras_negativas:
            # Comparo con la palabra actual
            if(j == k):
                # Reviso si ya existe registro de la palabra, si lo hay suma
                if(k in dic):
                    dic[k] += 1
                # Si la palabra no está, se añade con el contador a 1
                else:
                    dic[k] = 1

print(dic)
```
